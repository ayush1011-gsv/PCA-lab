# PCA (principal component analysis)

```python
import numpy as np
import pandas as pd
data={'x':[7,4,6,8,8,7,5,9,7,8],
      'y':[4,1,3,6,5,2,3,5,4,2]}
data=pd.DataFrame(data)
x_mean=data["x"].mean()
y_mean=data["y"].mean()

data["x-mean(x)"]=data['x']-x_mean
data["y-mean(y)"]=data['y']-y_mean

data["(x-mean(x))^2"]=(data["x"]-x_mean)**2
data["(y-mean(y))^2"]=(data['y']-y_mean)**2
data["(x-mean(x))(y-mean(y))"]=data["x-mean(x)"]*data['y-mean(y)']

variance_x=data["(x-mean(x))^2"].sum() / (data["(x-mean(x))^2"].count()-1)
variance_y=data["(y-mean(y))^2"].sum() / (data["(y-mean(y))^2"].count()-1)

covariance=data["(x-mean(x))(y-mean(y))"].sum() / (data["(x-mean(x))(y-mean(y))"].count() -1)
display(data)

print("\nvarince of x: ",variance_x,"\nvarince of y: ",variance_y)
print("\ncovarince: ",covariance)

covariance_matrix=np.array([[variance_x,covariance],[covariance,variance_y]])
print("\ncovarince matrix: \n",covariance_matrix)

eigenvalue,noraml_eigenvector=np.linalg.eig(covariance_matrix)
print("\neigenvalue: ")
print(eigenvalue)
print("\nnormal eigenvector")
print(noraml_eigenvector)

sorted_indices=np.argsort(eigenvalue)[::-1]
sorted_eigenvalue = eigenvalue[sorted_indices]
sorted_eigenvector=noraml_eigenvector[:,sorted_indices]# larger vecoter at 1st col
pc1_vector=-1*sorted_eigenvector[:,0]

print("\n 1st principle component vectoer :",pc1_vector)
centered_data=data[['x-mean(x)','y-mean(y)']].values
data["pc1"]=np.dot(centered_data,pc1_vector)
display(data[['x','y','pc1']])
```
